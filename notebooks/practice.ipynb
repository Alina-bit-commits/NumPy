```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Distribución normal
normal = np.random.normal(loc=50, scale=10, size=1000)

# Distribución uniforme
uniforme = np.random.uniform(low=0, high=100, size=1000)
```

```python
#Distribución Normal
print("Distribución Normal")
print("Media:", np.mean(normal))
print("Mediana:", np.median(normal))
print("Varianza:", np.var(normal))
print("Desvío estándar:", np.std(normal))
print("Mínimo:", np.min(normal))
print("Máximo:", np.max(normal))
```

```python
#Distribución Uniforme
print("Distribución Uniforme")
print("Media:", np.mean(uniforme))
print("Mediana:", np.median(uniforme))
print("Varianza:", np.var(uniforme))
print("Desvío estándar:", np.std(uniforme))
print("Mínimo:", np.min(uniforme))
print("Máximo:", np.max(uniforme))
```

```python

fig, ax = plt.subplots(1, 2, figsize=(12, 5))

# Histograma de la distribución normal
ax[0].hist(normal, bins=30, color="skyblue", edgecolor="black",
           label="Normal")
ax[0].set_title("Distribución Normal")
ax[0].set_xlabel("Valores")
ax[0].set_ylabel("Frecuencia")
ax[0].legend()
ax[0].grid(alpha=0.3)

# Histograma de la distribución uniforme
ax[1].hist(uniforme, bins=30, color="lightgreen", edgecolor="black",
           label="Uniforme")
ax[1].set_title("Distribución Uniforme")
ax[1].set_xlabel("Valores")
ax[1].set_ylabel("Frecuencia")
ax[1].legend()
ax[1].grid(alpha=0.3)

fig.suptitle("Comparación de Distribuciones Probabilísticas")

plt.tight_layout()
plt.show()
```

**Reflexión**

Al comparar ambas distribuciones, se observa que tienen medias similares, cercanas a 50, pero presentan diferencias en la dispersión de los datos.

En la distribución normal, la mayoría de los valores se concentran alrededor de la media, con un desvío estándar de 9,69. En cambio, en la distribución uniforme, los valores se distribuyen a lo largo del intervalo entre 0 y 100, con un desvío estándar mayor, de 29,28.

En los histogramas se puede observar que la distribución normal tiene una forma aproximada de campana, mientras que la uniforme presenta frecuencias relativamente similares a lo largo del intervalo.

En conclusión, aunque ambas distribuciones tienen promedios parecidos, la forma en que se distribuyen sus valores es diferente.
